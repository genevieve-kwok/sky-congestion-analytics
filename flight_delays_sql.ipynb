```python
import sqlite3
import pandas as pd
```

```python
conn = sqlite3.connect("flight_delays.db")

tables = pd.read_sql_query(
    "SELECT name FROM sqlite_master WHERE type='table';", conn
)
print("--- Tables in Database ---")
print(tables)

print("\n--- Schema Info ---")
schema_df = pd.read_sql_query("PRAGMA table_info(flight_info);", conn)
print(schema_df[['cid', 'name', 'type']])

row_count = pd.read_sql_query(
    "SELECT COUNT(*) as total_rows FROM flight_info;", conn
)
print(f"\nTotal rows imported: {row_count['total_rows'][0]:,}")
```

```python
# verifying the first 5 rows
sample_df = pd.read_sql_query(
    "SELECT * FROM flight_info LIMIT 5;", conn
)
print("--- Data Sample ---")
print(sample_df)

# checking for null values
null_check = pd.read_sql_query("""
SELECT
    COUNT(*) - COUNT(Flight_Number_Reporting_Airline) as missing_flight_nums, 
    COUNT(*) - COUNT(ArrDelay) as missing_arr_delays
FROM flight_info;
""", conn)

print("\n--- Missing Value Check ---")
print(null_check)
```
